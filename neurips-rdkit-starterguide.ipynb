  <p style="margin-bottom: 15px;">
    The dataframes uses <strong style="color: #e74c3c; font-weight: 600;">SMILES notation</strong> (Simplified Molecular Input Line Entry System) - a precise text-based method for representing molecular structures. SMILES not only shows the atoms that make up the compound but it also illustrates how the atoms are bonded and arranged in its Structure. Inorder for it to give such rich details it applies specific rules to describe molecular structures some of which are :
  </p>
    <div style="
    background: #f8f9fa;
    border-radius: 8px;
    padding: 15px;
    margin: 15px 0;
  ">
    <p style="margin-bottom: 10px; font-weight: 600;">SMILES Encoding Rules:</p>
    <ul style="padding-left: 20px; margin-bottom: 10px;">
      <li style="margin-bottom: 8px;">Atoms are represented by their element symbols (C for carbon, O for oxygen)</li>
      <li style="margin-bottom: 8px;">Single bonds are implied, double bonds shown by symbol "=" while triple bonds are shown by "#" </li>
      <li style="margin-bottom: 8px;">Branches are enclosed in parentheses</li>
      <li>Ring structures are indicated by matching numbers</li>
    </ul>
    <p style="margin: 0;">
      <strong>Example:</strong> <code style="background: #e8f4f8; padding: 4px 8px; border-radius: 4px; font-family: monospace; border: 1px solid #c1e0f0;">CCO</code> represents ethanol (CH₃-CH₂-OH) - two carbons single-bonded to each other, with the second carbon bonded to oxygen
    </p>
  </div>
  <p style="margin-bottom: 15px;">
    These dataframe also comprises of target polymer properties values that are to be predicted by our models , below is the description of these properties:
  </p>
  <div style="

        <td style="padding: 12px 2px; vertical-align: top; line-height: 1.5;">
          <strong>Glass Transition Temperature</strong>: The temperature at which polymer chains gain sufficient mobility to transition from rigid to flexible state. It's a crucial property that dictates a material's mechanical behavior, especially at different temperatures.

          <strong>Radius of Gyration</strong>: crucial parameter in polymer science, characterizing the extent to which a polymer chain is spread out. It represents the root-mean-square distance of the polymer's segments from its center of mass. This value helps quantify the size of a polymer coil, even when its shape is irregular due to folding and twisting.

       <p>In this section, we import all the essential libraries needed for data processing, feature generation, model training, and optimization. We also define a `Config` class that holds key  settings like random seeds, number of folds for cross-validation, training parameters, and paths for saving models and optimization results.</p>

<div style="
    background: #f8f9fa;
    border-radius: 8px;
    padding: 15px;
    margin-bottom: 15px;
  ">
   
<p style="margin: 8px 0px;">
      The <span style="color: #e74c3c; font-weight: 600;">get_canonical_smiles</span> function standardizes SMILES strings into their canonical form so that molecules with the same structure are recognized as identical. This is especially important when using GroupKFold, where we want to avoid splitting similar polymers across train and validation sets. By grouping on canonical SMILES, we ensure structurally identical polymers stay in the same fold.
    </p>
  </div>
  

</div>

```python
import numpy as np
import pandas as pd
from tqdm.notebook import tqdm
import optuna
import xgboost as xgb
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_absolute_error
from rdkit import Chem
from rdkit.Chem import Descriptors, Lipinski, rdMolDescriptors, Fragments, AllChem
from pathlib import Path
import logging
import warnings

# Completely silence all warnings and logging
warnings.filterwarnings('ignore')
optuna.logging.set_verbosity(optuna.logging.WARNING)
logging.getLogger('optuna').setLevel(logging.WARNING)

# ====================
# Configuration
# ====================
class Config:
    RANDOM_STATE = 42
    USE_GPU = True
    GPU_ID = 0
    N_FOLDS = 5
    MAX_ITERATIONS = 10000
    EARLY_STOPPING = 100
    OPTUNA_TRIALS = 100
    VERBOSE_EVAL = 100
    TARGET_COLUMNS = ["Tg", "FFV", "Tc", "Density", "Rg"]
    STUDY_DIR = Path("studies")
    MODEL_DIR = Path("models")
    
    def __init__(self):
        self.STUDY_DIR.mkdir(parents=True, exist_ok=True)
        self.MODEL_DIR.mkdir(parents=True, exist_ok=True)

    def get_canonical_smiles(smiles):
            """Convert SMILES to canonical form for consistent grouping"""
            try:
                mol = Chem.MolFromSmiles(smiles)
                if mol:
                    return Chem.MolToSmiles(mol, canonical=True)
                return smiles  # fallback to original if conversion fails
            except:
                return smiles
```

      <div style="flex: 1; min-width: 200px;">
        <p style="font-weight: 600; margin-bottom: 8px;">⚡ Polarity & Reactivity</p>
        <ul style="padding-left: 20px; margin: 0;">
          <li>LogP (hydrophobicity)</li>
          <li>Topological Polar Surface Area (TPSA)</li>
          <li>H-bond donors (HBD)</li>
          <li>H-bond acceptors (HBA)</li>
        </ul>
      </div>
      <div style="flex: 1; min-width: 200px;">
        <p style="font-weight: 600; margin-bottom: 8px;">🧪 Functional Groups</p>
        <ul style="padding-left: 20px; margin: 0;">
          <li>Ether count</li>
          <li>Ester count</li>
          <li>Amide count</li>
        </ul>
      </div>
    </div>
  </div>

  <p style="margin-bottom: 15px;">
    In addition to these descriptors, the function computes a <strong>binary Morgan fingerprint</strong> vector (default: <code>n_bits=1024</code>), which captures the presence or absence of substructures in the molecule using a circular hashing approach with configurable radius (<code>radius=2</code> by default).
  </p>

  <p style="margin-bottom: 15px;">
    Molecules that cannot be parsed or throw an error are safely skipped and excluded from the final dataset. The final output is a cleaned and enriched DataFrame that includes both the original data and the generated features — ready for modeling!
  </p>

</div>


```python
def generate_polymer_features(df, smiles_column='canonical_smiles', radius=2, n_bits=1024):
    def _calculate_features(smiles):
        try:
            mol = Chem.MolFromSmiles(smiles)
            if not mol:
                return None
            
            # 1. Calculate traditional descriptors
            descriptors = {
                'MW': Descriptors.MolWt(mol),
                'HeavyAtomCount': Descriptors.HeavyAtomCount(mol),
                'RotatableBonds': Descriptors.NumRotatableBonds(mol),
                'RingCount': rdMolDescriptors.CalcNumRings(mol),
                'LogP': Descriptors.MolLogP(mol),
                'TPSA': Descriptors.TPSA(mol),
                'HBD': Lipinski.NumHDonors(mol),
                'HBA': Lipinski.NumHAcceptors(mol),
                'EtherCount': Fragments.fr_ether(mol),
                'EsterCount': Fragments.fr_ester(mol),
                'AmideCount': Fragments.fr_amide(mol),
                'AromaticRingCount': rdMolDescriptors.CalcNumAromaticRings(mol),
                'BertzCT': Descriptors.BertzCT(mol),
                'BalabanJ': Descriptors.BalabanJ(mol),
            }
            
            # 2. Calculate Morgan fingerprints
            fp = AllChem.GetMorganFingerprintAsBitVect(mol, radius=radius, nBits=n_bits)
            fp_features = {f'morgan_{i}': int(bit) for i, bit in enumerate(fp)}
            
            # Combine both feature types
            return {**descriptors, **fp_features}
            
        except Exception as e:
            print(f"Skipping SMILES due to error: {smiles[:50]}... Error: {str(e)[:100]}...")
            return None

    features = []
    valid_indices = []
    
    print(f"Generating features (RDKit descriptors + {n_bits}-bit Morgan fingerprints)...")
    for idx, smi in tqdm(enumerate(df[smiles_column]), total=len(df)):
        feat = _calculate_features(smi)
        if feat is not None:
            features.append(feat)
            valid_indices.append(idx)
    
    features_df = pd.DataFrame(features)
    result_df = pd.concat([
        df.iloc[valid_indices].reset_index(drop=True),
        features_df.reset_index(drop=True)
    ], axis=1)
    
    print(f"Generated {len(features_df.columns)} total features")
    print(f"  - {14} RDKit descriptors")
    print(f"  - {n_bits} Morgan fingerprint bits")
    
    return result_df

train = generate_polymer_features(overall_train)

print("\n🎯 Generating molecular features for test data...")
test_data = generate_polymer_features(test)

# Identify feature columns
feature_columns = [col for col in train.columns 
                  if col not in Config.TARGET_COLUMNS + ['SMILES','canonical_smiles'] and not col.startswith('id')]

# print("\n🎯 Feature columns:", feature_columns)
```

<h1 id="XGBoost" style="

">
 5. 🎯 XGBoost Hyperparameter Optimization
</h1>

<div style="
  background: white;
  border-radius: 12px;
  border: 1px solid #e0e0e0;
  font-family: Arial, sans-serif;
  color: #333;
  padding: 18px;
  font-size: 15px;
  line-height: 1.6;
  margin-bottom: 25px;
">
  <p style="margin-bottom: 15px;">
    We implement an optimization strategy using Optuna to automatically find the best XGBoost hyperparameters for each polymer property prediction task. This approach replaces fixed parameters with values that adapts to each target's specific characteristics.
  </p>

  <p style="margin-bottom: 15px;">
    The process begins with a base configuration containing sensible defaults for regression tasks, including GPU acceleration support for faster computation. All optimization logic is encapsulated in an <span style="color: #e74c3c; font-weight: 600;">objective</span> function that Optuna repeatedly evaluates with different parameter combinations.
  </p>

  <div style="
    background: #f8f9fa;
    border-radius: 8px;
    padding: 15px;
    margin-bottom: 15px;
  ">
    <p style="margin-bottom: 8px;">
      We employ <strong>GroupKFold cross-validation</strong> with SMILES strings as grouping identifiers. This ensures that:
    </p>
    <ul style="padding-left: 20px; margin-bottom: 0;">
      <li style="margin-bottom: 5px;">All instances of the same polymer stay within the same fold</li>
      <li>No information leaks between training and validation sets</li>
    </ul>
  </div>




```python
def get_xgb_params(Config):
    """Get base XGBoost parameters"""
    params = {
        'objective': 'reg:squarederror',
        'eval_metric': 'mae',
        'seed': Config.RANDOM_STATE,
        'verbosity': 0
    }
    
    if Config.USE_GPU:
        params.update({
            'tree_method': 'gpu_hist',
            'gpu_id': Config.GPU_ID,
            'predictor': 'gpu_predictor'
        })
    return params
    

def objective(trial, X, y, groups, Config):
    """Optuna optimization objective function"""
    params = get_xgb_params(Config)
    params.update({
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.1, log=True),
        'max_depth': trial.suggest_int('max_depth', 5, 20),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'gamma': trial.suggest_float('gamma', 1e-8, 1.0, log=True),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-8, 1.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-8, 2.0, log=True),
        'min_child_weight': trial.suggest_int('min_child_weight', 1, 10),
    })

    group_kfold = GroupKFold(n_splits=Config.N_FOLDS)
    cv_scores = []

    for train_idx, valid_idx in group_kfold.split(X, y, groups=groups):
        dtrain = xgb.DMatrix(X[train_idx], label=y[train_idx])
        dvalid = xgb.DMatrix(X[valid_idx], label=y[valid_idx])

        bst = xgb.train(
            params,
            dtrain,
            num_boost_round=Config.MAX_ITERATIONS,
            evals=[(dtrain, 'train'), (dvalid, 'valid')],
            early_stopping_rounds=Config.EARLY_STOPPING,
            verbose_eval=False
        )
        cv_scores.append(bst.best_score)

    return np.mean(cv_scores)
```


<!-- Target-wise Training Header -->
<h1 id="Target"
    style="
    background: linear-gradient(135deg, #0f2027, #203a43, #2c5364);
  color: white;
  padding: 12px 24px;

  margin: 20px 0;
">
  6. 🎯 Target-Specific Model Training with Optimization
</h1>

<div style="
  background: white;
  border-radius: 12px;
  border: 1px solid #e0e0e0;
  font-family: Arial, sans-serif;
  color: #333;
  padding: 18px;
  font-size: 15px;
  line-height: 1.6;
  margin-bottom: 25px;
">
  <p style="margin-bottom: 15px;">
   This section defines how we train a separate XGBoost model for each target polymer property (like Tg, Tc, FFV, etc.). Instead of  training a model with fixed parameters, we use Optuna to fine-tune hyperparameters specifically for each target. This is done through the following processes:
  </p>

  <div style="
    background: #f8f9fa;
    border-radius: 8px;
    padding: 15px;
    margin-bottom: 15px;
  ">
    <ol style="padding-left: 20px; margin-bottom: 0;">
      <li style="margin-bottom: 10px;">
        <strong>Hyperparameter Optimization</strong> (via Optuna):
        <ul style="padding-left: 20px; margin-top: 8px;">
          <li>Evaluated through 5-fold grouped cross-validation</li>
        </ul>
      </li>
      <li style="margin-bottom: 10px;">
        <strong>Final Model Training</strong>:
        <ul style="padding-left: 20px; margin-top: 8px;">
          <li style="margin-bottom: 5px;">Uses best parameters from optimization</li>
          <li>Trains on 100% of available data</li>
        </ul>
      </li>
      <li style="margin-bottom: 10px;">
        <strong>Performance Assessment</strong>:
        <ul style="padding-left: 20px; margin-top: 8px;">
          <li style="margin-bottom: 5px;">Training set metrics (MAE)</li>  
        </ul>
      </li>
    </ol>
  </div>
</div>

```python
def train_single_target(X, y, groups, target_name, Config):
    """Train model for a single target property"""
    print(f"\nStarting optimization for {target_name}")
    
    study = optuna.create_study(
        direction='minimize',
        sampler=optuna.samplers.TPESampler(seed=Config.RANDOM_STATE),
        pruner=optuna.pruners.MedianPruner(n_warmup_steps=10)
    )
    
    study.optimize(
        lambda trial: objective(trial, X, y, groups, Config),
        n_trials=Config.OPTUNA_TRIALS,
        show_progress_bar=True
    )

    best_params = get_xgb_params(Config)
    best_params.update(study.best_params)
    
    print(f"Best MAE for {target_name}: {study.best_value:.4f}")
    
    dtrain = xgb.DMatrix(X, label=y)
    model = xgb.train(
        best_params,
        dtrain,
        num_boost_round=Config.MAX_ITERATIONS,
        verbose_eval=Config.VERBOSE_EVAL
    )

    # Evaluate on training set
    preds = model.predict(dtrain)
    train_mae = mean_absolute_error(y, preds)
    
    return model, {
        'best_cv_mae': study.best_value,
        'train_mae': train_mae,
        'params': best_params,
        'n_samples': len(y)
    }
```

  <p style="margin-bottom: 15px;">
    With our features prepared and model training function defined, we now loop through each target property (Tg, FFV, Tc, Density, Rg) and train a dedicated model for each.Below is the Process Flow:
  </p>

  <div style="
    background: #f8f9fa;
    border-radius: 8px;
    padding: 15px;
    margin-bottom: 15px;
  ">
    <ol style="padding-left: 20px; margin-bottom: 0;">
      <li style="margin-bottom: 10px;">
        <strong>Data Preparation</strong>:
        <ul style="padding-left: 20px; margin-top: 8px;">
          <li style="margin-bottom: 5px;">filtering out missing values specific to the target and extracting the corresponding features and group labels</li>
        </ul>
      </li>
      <li style="margin-bottom: 10px;">
        <strong>Model Training</strong>:
        <ul style="padding-left: 20px; margin-top: 8px;">
          <li>call the <span style="color: #e74c3c; font-weight: 600;">train_single_target</span> function for each target, which handles hyperparameter tuning and model training using XGBoost and Optuna.</li>
        </ul>
      </li>
      <li style="margin-bottom: 10px;">
        <strong>Model Storage</strong>:
        <ul style="padding-left: 20px; margin-top: 8px;">
          <li>Each trained model is stored in a dictionary for future use, and saved to disk in JSON format to allow reuse without retraining</li>
        </ul>
      </li>
      <li>
        <strong>Performance Reporting</strong>:
        <ul style="padding-left: 20px; margin-top: 8px;">
          <li>we print a concise summary showing how well each model performed on its training data and during cross-validation </li>
        </ul>
      </li>
    </ol>
  </div>

  
</div>

```python
# Prepare training data
X_train = train[feature_columns].values
y_df = train[Config.TARGET_COLUMNS]
groups = train['canonical_smiles'].factorize()[0]

# Initialize storage for models and results
models = {}
results = {}

# Ensure models directory exists
Config.MODEL_DIR.mkdir(parents=True, exist_ok=True)

# Train models for each target
for target in Config.TARGET_COLUMNS:
    if target not in y_df.columns:
        print(f"Warning: Target {target} not found in data")
        continue
        
    mask = y_df[target].notna()
    if mask.sum() == 0:
        print(f"Warning: No valid samples for target {target}")
        continue
    
    X_target = X_train[mask]
    y_target = y_df[target][mask].values
    groups_target = groups[mask]

    try:
        model, target_results = train_single_target(
            X_target, y_target, groups_target, target, Config
        )
        
        models[target] = model
        results[target] = target_results
        
        # Save model
        model_path = Config.MODEL_DIR / f"{target}_model.json"
        model.save_model(str(model_path))
        
        print(f"✅ {target}: CV MAE={target_results['best_cv_mae']:.4f}, Train MAE={target_results['train_mae']:.4f}")
        
    except Exception as e:
        print(f"Failed to train {target}: {e}")
        continue

# Print summary
print("\n" + "="*60)
print("TRAINING SUMMARY")
print("="*60)

for target, result in results.items():
    print(f"\n🎯 {target}:")
    print(f"  Samples: {result['n_samples']}")
    print(f"  CV MAE: {result['best_cv_mae']:.4f}")
    print(f"  Train MAE: {result['train_mae']:.4f}")
```

After training, we use the saved models to generate predictions for the test set. For each target property, we apply the corresponding model to the test feature matrix. The predictions for all targets are then compiled into a single submission DataFrame, preserving the required column structure with id, Tg, FFV, Tc, Density, and Rg and then submitting  </p>

```python
# Get same feature columns used in training
X_test = test_data[feature_columns].values
print(f"Test data shape: {X_test.shape}")

def generate_predictions(models, X_test, test_ids):
    """Generate predictions for all targets using trained models"""
    predictions = {}
    
    for target, model in models.items():
        try:
            dtest = xgb.DMatrix(X_test)
            predictions[target] = model.predict(dtest)
            print(f"Generated predictions for {target}")
        except Exception as e:
            print(f"Prediction failed for {target}: {e}")
            predictions[target] = np.zeros(len(X_test))  # Default to zeros if prediction fails
    
    # Create DataFrame with IDs and predictions
    submission = pd.DataFrame({'id': test_ids})
    for target in Config.TARGET_COLUMNS:
        if target in predictions:
            submission[target] = predictions[target]
        else:
            print(f"Warning: No predictions for {target}, filling with zeros")
            submission[target] = 0.0
    
    # Ensure correct column order
    submission = submission[['id', 'Tg', 'FFV', 'Tc', 'Density', 'Rg']]
    return submission

# Generate predictions
print("\nGenerating predictions...")
submission = generate_predictions(models, X_test, test['id'])

# Save submission file
submission_path = "submission.csv"
submission.to_csv(submission_path, index=False)
print(f"\nSubmission file saved to {submission_path}")

# Show sample of predictions
print("\nSample predictions:")
print(submission.head())

```
